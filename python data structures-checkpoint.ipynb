```python
#string op
str='Afifa'
print(str.isalnum())#check if all char are nums "afifa123".isalnum() 

# True  (letters + numbers)
"12345".isalnum()        # True  (only numbers)
"afifa_123".isalnum()    # False (_ is NOT alphanumeric)
"afifa 123".isalnum()    # False (space is NOT allowed)
"afifa@".isalnum()       # False (@ is special character)
"".isalnum()             # False (empty string)

print(str.isupper())
print(str.islower())
print(str.isalpha())#check all are alphabets
print(str.isspace())
print(str.isdigit())
print(str.istitle())#check if ther is title words
print(str.startswith('a'))
print(str.endswith('fa'))
```
